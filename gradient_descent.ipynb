Batch Gradient Descent Using OOP

```python
import numpy as np

class LinearRegression:
    
    def __init__(self, learning_rate=0.01, epochs=1000):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.theta0 = 0
        self.theta1 = 0

    def fit(self, X, y):
        n = len(X)

        for i in range(self.epochs):
            
            # Predictions for all observations
            y_pred = self.theta0 + self.theta1 * X

            # Calculate gradients using Mean Squared Error loss function
            d_theta0 = (-2/n) * np.sum(y - y_pred)
            d_theta1 = (-2/n) * np.sum(X * (y - y_pred))

            # Update parameters
            self.theta0 = self.theta0 - self.learning_rate * d_theta0
            self.theta1 = self.theta1 - self.learning_rate * d_theta1

    def predict(self, X):
        return self.theta0 + self.theta1 * X


```

```python

X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])
model = LinearRegression(learning_rate=0.01, epochs=1000)

model.fit(X, y)

# Display estimated parameters
print("Theta 0:", model.theta0)
print("Theta 1:", model.theta1)

# Prediction
X_new = np.array([6, 7])
print("Predictions:", model.predict(X_new))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.scatter(X, y, label="Actual Data")

plt.plot(X, model.predict(X), label="Regression Line")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression using Batch Gradient Descent")

plt.legend()
plt.grid(True)
plt.show()
```

 Mini Batch Gradient Descent Using OOP

```python
class MiniBatchGradientDescent:

    def __init__(self, learning_rate=0.01, epochs=1000, batch_size=2):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.batch_size = batch_size
        self.theta0 = 0
        self.theta1 = 0

    def fit(self, X, y):

        n = len(X)

        for epoch in range(self.epochs):
            indices = np.random.permutation(n)
            X_shuffled = X[indices]
            y_shuffled = y[indices]

            # Divide data into mini-batches
            for i in range(0, n, self.batch_size):

                X_batch = X_shuffled[i:i + self.batch_size]
                y_batch = y_shuffled[i:i + self.batch_size]

                
                y_pred = self.theta0 + self.theta1 * X_batch
                batch_n = len(X_batch)

                # Calculate gradients
                d_theta0 = (-2 / batch_n) * np.sum(y_batch - y_pred)

                d_theta1 = (-2 / batch_n) * np.sum(
                    X_batch * (y_batch - y_pred)
                )
                self.theta0 = self.theta0 - self.learning_rate * d_theta0

                self.theta1 = self.theta1 - self.learning_rate * d_theta1

    def predict(self, X):
        return self.theta0 + self.theta1 * X

```

```python
X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

# Create model
model = MiniBatchGradientDescent(
    learning_rate=0.01,
    epochs=1000,
    batch_size=2
)
# Train model
model.fit(X, y)
# Display estimated parameters
print("Theta 0:", model.theta0)
print("Theta 1:", model.theta1)

X_new = np.array([6, 7])
print("Predictions:", model.predict(X_new))
```

```python
# Plot 
plt.figure(figsize=(8, 5))
plt.scatter(X, y, label="Actual Data")
plt.plot(
    X,
    model.predict(X),
    label="Regression Line"
)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Mini-Batch Gradient Descent")

plt.legend()
plt.grid(True)

plt.show()
```

Stochastic Gradient Descent Using OOP

```python
class StochasticGradientDescent:

    def __init__(self, learning_rate=0.01, epochs=1000):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.theta0 = 0
        self.theta1 = 0

    def fit(self, X, y):

        n = len(X)

        for epoch in range(self.epochs):
            indices = np.random.permutation(n)

            X_shuffled = X[indices]
            y_shuffled = y[indices]

            for i in range(n):

                X_i = X_shuffled[i]
                y_i = y_shuffled[i]
                y_pred = self.theta0 + self.theta1 * X_i

                # Calculate gradients
                d_theta0 = -2 * (y_i - y_pred)

                d_theta1 = -2 * X_i * (y_i - y_pred)

                
                self.theta0 = self.theta0 - self.learning_rate * d_theta0
                self.theta1 = self.theta1 - self.learning_rate * d_theta1

    def predict(self, X):
        return self.theta0 + self.theta1 * X
```

```python
X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])

# Create model
model = StochasticGradientDescent(
    learning_rate=0.01,
    epochs=1000
)

model.fit(X, y)


# Display estimated parameters
print("Theta 0:", model.theta0)
print("Theta 1:", model.theta1)


X_new = np.array([6, 7])

print("Predictions:", model.predict(X_new))


```

```python
# Plot 
plt.figure(figsize=(8, 5))
plt.scatter(X, y, label="Actual Data")
plt.plot(
    X,
    model.predict(X),
    label="Regression Line"
)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Stochastic Gradient Descent")

plt.legend()
plt.grid(True)

plt.show()
```

Gradient Descent is an optimization algorithm used to minimize the loss function by repeatedly updating model parameters in the direction of the negative gradient.

Batch Gradient Descent uses all training observations to calculate the gradient and then updates the parameters once.It is stable but can be computationally expensive for large datasets.

Stochastic Gradient Descent uses one observation at a time to calculate the gradient and immediately updates the parameters.It is faster and can work well with large datasets, but the updates can be noisy.

Mini-Batch Gradient Descent uses a small group of observations to calculate the gradient before updating the parameters.It provides a balance between the stability of Batch GD and the speed of SGD.
