```python
# Page Rank Algorithm

# Number of webpages
N = int(input("Enter number of webpages: "))

# Damping factor
d = 0.85

# Enter links
links = {}

for i in range(N):
    page = input("Enter page name: ")
    outgoing = input("Enter pages linked from " + page +
                     " (separated by space, enter 'none' if no link): ")

    if outgoing.lower() == "none":
        links[page] = []
    else:
        links[page] = outgoing.split()


# Initial PageRank
pages = list(links.keys())

PR = {}

for page in pages:
    PR[page] = 1 / N

print("\nInitial PageRank:")
for page in pages:
    print(page, "=", PR[page])


# Number of iterations
iterations = int(input("\nEnter number of iterations: "))


# PageRank calculation
for iteration in range(iterations):

    new_PR = {}

    for page in pages:

        # Teleportation/random jump component
        rank = (1 - d) / N

        # Find pages pointing to the current page
        for other_page in pages:

            if page in links[other_page]:

                # Number of outgoing links
                L = len(links[other_page])

                # Ignore dangling nodes
                if L > 0:
                    rank += d * PR[other_page] / L

        new_PR[page] = rank

    PR = new_PR

    print("\nIteration", iteration + 1)

    for page in pages:
        print(page, "=", round(PR[page], 6))


# Final PageRank
print("\nFinal PageRank:")
for page in pages:
    print(page, "=", round(PR[page], 6))
```
